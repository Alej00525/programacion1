## Instruccion 1

```python
# Diferencia

A = {1,2,3,4}
B = {3,4,5,6}
A.difference(B)
```

La diferencia entre dos conjuntos muestra los elementos de A que no comparte con B
el conjunto A tiene los elementos {1,2,3,4} y el B los elementos {3,4,5,6} el conjunto resultante sera {1,2} que son los elementos de a que NO estan en B

```python
# Diferencia simetrica

A = {1,2,3,4}
B = {3,4,5,6}
A.symmetric_difference(B)
```

La diferencia simetrica entre dos conjuntos muestra los elementos de A que no comparte con B y los elementos de B que no comparte con A en un solo conjunto
el conjunto A tiene los elementos {1,2,3,4} y el B los elementos {3,4,5,6} el conjunto resultante sera {1,2,5,6} que son los elementos de a que NO comparten entre si.

##Instruccion 2


```python
# Is super set

A = {1,2,3,4,5,6}
B = {1,2,3}
A.issuperset(B)
```

El metodo verifica si el conjunto B esta totalmente contenido dentro de A, si es cierto entrega TRUE si es falso FALSE
como todos los elementos de B estan en A el resultado es TRUE

```python
# Is disjoint

A = {1,2,3,4,5,6}
B = {1,2,3}
A.isdisjoint(B)
```

El metodo verifica si NO existe interseccion entre los conjuntos es decir que si no comparten ningun elemento (se puede leer como "¿los conjuntos A y B estan desconectados?")
como A y B comparten elementos el resultado es FALSE

## Instruccion 3

```python
# .update

datos2 = {'a':1,'b':2,'c':3}
datos2.update({'c':4})
datos2.update({'d':3})
datos2.update({'a':1})
print(datos2)
```

El metodo update actualiza una clave:valor dentro del diccionario, si la clave ya esta en el diccionario, modifica solo el valor, si la clave NO esta en el diccionario añade la pareja clave:valor

si se quiere añadir una pareja clave:valor que ya esta simplemente no hace nada

##Instruccion 4

```python
# setdefault
datos3 = {'a':1,'b':2,'c':3,}
datos3.setdefault('d',4)
datos3.setdefault('b',4)
datos3.setdefault('a',1)
print(datos3)
```

El metodo setdefault permite añadir una clave que tenga como valor por defecto NONE (sin valor) y si se le aplica a claves que ya estan en el diccionario simplemente no hara nada y entregara el valor que ya tiene asignada esa clave

## Codigo 1

```python
def cnnumeros(m):
 if m is not str(m):
  if isinstance(m, int) and m >= 0:  # pregunta si m es entero y mayor o igual que 0
   for n in range(0, (m+1)):
    if n == 1 or n == 0:
     print("El número es ", n)
    else:
     for x in range(2, n): # para cuando el numero llegue a 2 empiece a factorizar o a decir si es primo
       if n % x == 0:
         print(n, 'es igual a', x, '*', n // x)
         break
     else: # Este else corresponde al for, no al if
      print(n, 'es un número primo')
  elif m is float(m): # si el numero es decimal
   print("El número no es entero")
  else: # si el numero es negativo
   print("El número es negativo")
 else: # si el numero es una cadena (tambien se puede hacer con un try, except)
   print("lo que ingresaste no es un numero!!")
```

```python
cnnumeros(3.8)
```

## Codigo 2

```python
def contadorcaracteres(texto):
 vocalesmin = "aeiou"
 vocalesmay = vocalesmin.upper()
 consonantesmin = "bcdfghjklmnñpqrstvwxyz"
 consonantesmay = consonantesmin.upper()
 numeros = "0123456789"
 espacio = " "
 contador0 = 0
 contador1 = 0
 contador2 = 0
 contador3 = 0
 contador4 = 0
 contador5 = 0
 contador6 = 0    # un contador unico para cada uno de los caracteres que se van a diferenciar

 for caracter in texto:
  if caracter in vocalesmin:
   contador0 += 1
  elif caracter in vocalesmay:
   contador1 += 1
  elif caracter in consonantesmin:
   contador2 += 1
  elif caracter in consonantesmay:
   contador3 += 1
  elif caracter in numeros:
   contador4 += 1
  elif caracter in espacio:
   contador5 += 1
  else: # el programa comprueba a que categoria pertenece cada caracter y le suma una unidad al contador
   contador6 += 1
  # el programa crea un diccionario con la cantidad de caracteres que pertenece a cada categoria
 return({"Cantidad de vocales minusculas": contador0,
 "Cantidad de vocales mayusculas": contador1,
 "Cantidad de consonantes minusculas": contador2,
 "Cantidad de consonantes mayusculas": contador3,
 "Cantidad de numeros": contador4,
 "Cantidad de espacios": contador5,
 "Cantidad de simbolos": contador6})
```

```python
contadorcaracteres(str(input("ingrese un texto: ")))
```
