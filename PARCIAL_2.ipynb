## PARTE III

```python
PRUEBA_A=np.random.normal(loc=7.4, scale=0.4, size=55)
PRUEBA_B=np.random.gamma(shape=306, scale=0.0229, size=55)
```

```python
notas = pd.DataFrame({
    'A':PRUEBA_A,
    'B':PRUEBA_B
})
notas
```

```python
print('Analisis prueba A')
print(notas['A'].describe())
print('')
print('Analisis prueba B')
print(notas['B'].describe())
```

```python
from scipy import stats
```

```python
print('Prueba de shapiro para la prueba A')
print(stats.shapiro(notas['A']))
print('')
print('Prueba de shapiro para la prueba B')
print(stats.shapiro(notas['B']))
```

En la prueba A se dice que estan distribuidos normalmente con una confianza del 73% (valor p > a 0.05)

mientras que en la prueba B se da con una confianza del 38% (valor p > a 0.05)

Es decir, ninguna arroja una confianza del 95%

```python
t_stat, pt_val = stats.ttest_rel(notas["A"], notas['B'])
print("Estadístico t:", t_stat)
print("Valor p:", pt_val)
```

como el valor p es mayor a 0.05 podemos concluir que los resultados de la prueba no tuvieron diferencias significativas con una confianza del 95% es decir que estan relacionadas

```python
k_stat, pk_val = stats.kendalltau(notas["A"], notas["B"])

print("Coeficiente de correlacion K:", k_stat)
print("Valor p:", pk_val)
```

como el valor p es mayor a 0.05 podemos concluir que los resultados de la prueba no tuvieron diferencias significativas con una confianza del 95% es decir que estan relacionadas

```python
s_stat, ps_val = stats.spearmanr(notas["A"], notas["B"])

print("Coeficiente de correlacion S:", s_stat)
print("Valor p:", ps_val)
```

como el valor p es mayor a 0.05 podemos concluir que los resultados de la prueba no tuvieron diferencias significativas con una confianza del 95% es decir que estan relacionadas
