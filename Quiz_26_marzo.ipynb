```python
%%writefile estadistica.py
import statistics
def calcular_promedio(n0):
  return (sum(n0))/len(n0)

def calcular_mediana(n0):    # para todos n0 es una lista
  return statistics.median(n0)

def calcular_moda (n0):
  return statistics.mode(n0)
```

```python
%%writefile estudiante.py
def pedir_nombre():
  return input("Ingrese el nombre del estudiante: ")

def pedir_notas(n):
  notas = []
  for a in range (n):
    notas.append(float(input(f"Ingrese la nota {a+1}: ")))

  return notas
```
