```python
# Ejemplo de conversión de distintos tipos a False

print(0, "->", bool(0))
print(0.0, "->", bool(0.0))
print("", "->", bool(""))
print(list(), "->", bool(list()))
print(tuple(), "->", bool(tuple()))
print(set(), "->", bool(set()))
print(dict(), "->", bool(dict()))
print(None, "->", bool(None))

# Bool(ELEMENTO) sera falso siempre que el elemento corresponda al valor "nulo" de su categoria
```

```python
# Ejemplo de conversión de distintos tipos a True

print(333, "->", bool(333))
print(5.25, "->", bool(5.25))
print("letras", "->", bool("letras"))
print(list((1, 2)), "->", bool(list((1, 2))))
print(tuple((1, 2)), "->", bool(tuple((1, 2))))
print(set((1, 2)), "->", bool(set((1, 2))))
print(dict([(1,2)]), "->", bool(dict([(1,1)])))

# Bool(ELEMENTO) sera verdadero siempre que el elemento corresponda a algun valor diferente al "nulo" de su categoria
```

```python
# Ejemplo if-else

sisben = str(input("Introduzca su grupo sisben (A, B, C, D): "))
if sisben == "A":                             # si cumple la condicion ejecute...
  print("Muy pobre :(")
else:                                         # si no...
  print("No tan pobre jaja")
```

```python
#Ejemplo if-elif-else

nota = float(input("Introduzca su nota: "))
if 0 <= nota and nota < 3:
  print("Tontisimo")
elif 3 <= nota and nota < 3.8:
  print("Puede mejorar")
elif 3.8 <= nota and nota < 4.2:
  print("Muy bien")
elif 4.2 <= nota and nota <= 5:
  print("Super Excelente")
else:
  print("Nota fuera de rango")

  # si se cuple entonces...
  # sino entonces...
  # si no se cumplio nada entonces...
```

```python
# Ejemplo while

contador = 3
while contador != 0:    # mientras se cumpla una condicion ejecuta el script
  contador = contador - 1
  print("El contenido de contador es", contador)
```

```python
# for sobre estructuras de datos

lista = list((2, 1, 0))
tupla = tuple((2, 1, 0))
conjunto = set((2, 1, 0))
diccionario = dict((("epa","Dos"), (1, "Uno"), (0, "Cero")))

for iterador in diccionario: # aqui el iterador toma el valor de la clave
  print(iterador)
  print(diccionario[iterador])
```

```python
# Acceso por índices a una lista con un for

lista = ["Vamos", "a", "acceder", "a", "esta", "lista", "por", "índices"]

for indice in range(len(lista)):  # vamos a darle a indice un valor EN el rango que abarca la longitud de la lista y ejecuta el codigo con ese valor para el indice
  print("Indice:", indice,"Elemento:", lista[indice]) # al terminar el codigo lo repite con el siguiente valor para el indice
```

```python
# for para visualizar el efecto de las sentencias break y continue

for numero in range(5): # simplemente ejecuta el codigo con 0, 1, 2, 3, 4
  print(numero)
```

```python
# Visualización del efecto de la sentencia break

for numero in range(5):  # ejecuta el codigo con 0, 1, 2, 3, 4 PERO cuando llega a 3 este se detiene y no hace nada mas
  if numero == 3:
    break
  print(numero)
```

```python
# Visualización del efecto de la sentencia continue

for numero in range(5): # ejecuta el codigo con 0, 1, 2, 3, 4 PERO cuando llega a 3 este omite el paso y sigue para 4 como si nada
  if numero == 3:
    continue
  print(numero)
```
