```python
usuarios = {
    "ANA": "1234",
    "JUAN": "abcd"
}

try:
    usuario = input("Usuario: ").upper()
    password = input("Contraseña: ")

    if usuario not in usuarios:
        raise KeyError("El usuario no existe")

    if usuarios[usuario] != password:
        raise ValueError("Contraseña incorrecta")

    else:
        print("Bienvenido,", usuario)

except KeyError as e:
    print("Error:", e)

except ValueError as e:
    print("Error:", e)
```

a pesar de decir '5'/'0' que deberia dar ZeroDivisionError nos entrega un error Type esto debido a que no trasnformarmos las cadenas a numeros para operar y como en las cadenas no esta definida la division pues genera error

```python
# IA
try:
    # El programa se detendrá aquí mismo si el usuario ingresa una letra
    num1 = float(input("Ingrese el primer número: "))

    # Si lo anterior falla, el código de abajo nunca se ejecutará
    num2 = float(input("Ingrese el segundo número: "))
    op = input("Ingrese la operación (+, -, *, /): ")

    if op == "+":
        resultado = num1 + num2
    elif op == "-":
        resultado = num1 - num2
    elif op == "*":
        resultado = num1 * num2
    elif op == "/":
        # Python lanza ZeroDivisionError automáticamente,
        # pero es bueno validarlo o dejar que el except lo atrape
        resultado = num1 / num2
    else:
        raise ValueError("Operación no válida")

    print("El resultado es:", resultado)

except ValueError:
    # Este bloque atrapa tanto el error de letras en los números
    # como el error de "Operación no válida"
    print("Error: Entrada no válida. Asegúrese de ingresar números y un operador correcto.")

except ZeroDivisionError:
    print("Error: No se puede dividir entre cero.")

except Exception as e:
    # Atrapa cualquier otro error inesperado
    print(f"Ocurrió un error inesperado: {e}")
```

```python
try:

    datos = input("Ingresa notas separadas por coma: ")

    notas = [float(n) for n in datos.split(",")]

    if len(notas) == 0:
        raise ValueError("No ingresaste notas")

    promedio = sum(notas) / len(notas)

    posicion = int(input("¿Qué nota quieres ver (índice)?: "))

    if posicion < 0:
        raise ValueError("Ingresa solo índices positivos")

    print("Nota seleccionada:", notas[posicion])

    print("Promedio:", promedio)

except ValueError as e:
    print("Error de valor:", e)

except ZeroDivisionError:
    print("No se puede dividir entre cero")

except IndexError:
    print("Índice fuera de rango")
```
