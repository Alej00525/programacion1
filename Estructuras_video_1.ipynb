```python
# Listas

lista = [5, True, "25", "PARRA"]

print(lista)          # Imprime la lista
print(lista[-1])      # Imprime el ultimo elemente de la lista
print(lista[1:3])     # Imprime los indices 1 y 2

lista[2] = [3, 2, 1, 'MODIFICADO'] # Se pueden modificar indices individuales, tambien se pueden meter listas dentro de listas
print(lista)

print(len(lista)) # El metodo len nos dice cuantos elementos tiene la lista principal
```

```python
lista_nueva = [1, 2, 3, 4, 5]

lista_nueva.append(3)         # Añade un elemento al final

print(lista_nueva)

print(lista_nueva.count(3))   # Cuenta las veces que se repite un elemento

print(lista_nueva.index(4))   # Nos dice en que indice aparece un elemento por primera vez

lista_nueva.remove(3)         # Elimina un elemento la primera vez que aparece

print(lista_nueva)
```

```python
tupla = ("¿Te gusta el queso?", True, False)
print(tupla)

print(tupla[0])             # Para llamar los elementos de una tupla se utiliza un indice
print(tupla[1])
print(tupla[2])
print()

print(tupla.count(True))    # las tuplas solo tienen dos metodos, count e index
print(tupla.index(False))
```

```python
print(set())

print(set([5, 2, 5, 1, 1.5]))       # Se puede transformar una list en un conjunto
print(set((5, 2, 5, 1, 1.5)))       # Se puede transformar una tupla en un conjunto
print(set("52511.5"))               # se puede transformar una cadena en un conjunto (cada caracter entendido por elemento)

conjunto = set([2, 3, 3, 4])
conjunto_2 = set([5, 3, 5, 6])      # No existe repeticion en los conjuntos, y no tiene un orden predefinido
conjunto_3 = set([4, 2])

print(conjunto)
print(conjunto_2)
print(conjunto_3)

conjunto.add(1)           # El metodo add añade un elemento
print(conjunto)

conjunto.remove(1)        # El metodo remove elimina un elemento
print(conjunto)

print(conjunto.intersection(conjunto_2))    # El metodo interseccion nos entrega los elementos que estan en ambos conjuntos
print(conjunto_2.issubset(conjunto))        # El metodo issubset nos dice si un conjunto completo esta dentro de otro
print(conjunto_3.issubset(conjunto))
```

```python
diccionario = {1: "Uno", 2: "Dos"}
print(diccionario)
diccionario[3] = "Tres"         # se pueden añadir claves valor de la forma (diccionario(clave) = valor)
print(diccionario)

dict_lista_tuplas = dict([(1, "Uno"), (2, "Dos"), (3, "Tres")]) # se pueden crear diccionarios con listas o tuplas que adentro tienen listas o tuplas, de la forma (clave, valor)
print(dict_lista_tuplas)

dict_lista_string = dict(Uno = 1, Dos = 2, Tres = 3) # si se aplica dict a un script de la fomra (variable = valor) transforma la variable en la clave y el valor lo deja como valor
print(dict_lista_string)

dict_tipos = {1: "integer",
              2.2: "float",
              "texto": "string",
              (1, 2): "tupla"}

print(dict_tipos)

dict_repeticion = {1: "Primero", 1: "Último"} # si se repite una clave, a esta se le asignara el valor de la ultima vez que se reppite
print(dict_repeticion)

print(diccionario,
      diccionario.keys(),         # nos entrega las claves
      diccionario.values(),       # nos entrega los valores
      diccionario.items())        # nos entrega tuplas de la forma (clave,valor)

claves = diccionario.values()
print(claves)
diccionario[1] = "One"   # se puede modificar una clave pre-existente de la forma (diccionario(clave vieja) = nuevo valor)
print(claves)
diccionario.pop(2)  # elimina la clave y su valor
print(diccionario)
```
